**Ramesh’s basic salary is input through the keyboard. His dearness allowance is 40% of basic salary, and
house rent allowance is 20% of basic salary. Write a program to calculate his gross salary.**

```python
basic_salary=eval(input('enter salary:'))
dearness_allow=basic_salary*40/100
rent_allow=basic_salary*20/100
gross_salary= basic_salary + dearness_allow + rent_allow
print(f'Gross Salaryb is {gross_salary}')
```

**The length & breadth of a rectangle and radius of a circle are input through the keyboard. Write a program
to calculate the area & perimeter of the rectangle, and the area & circumference of the circle.**

```python
length = eval(input('enter length:'))
breadth = eval(input('enter breadth:'))
radius = eval(input('enter radius:'))
area_of_rectangle= length * breadth
perimeter_of_rectangle= 2*length*breadth
area_of_circle = 3.14*radius*radius
circumference_of_circle = 2*3.14*radius
print(f'area of rectangle {area_of_rectangle}')
print(f'perimetre of rectangle {perimeter_of_rectangle}')
print(f'area of circle {area_of_circle}')
print(f'circumference of circle {circumference_of_circle}')
```

**7. If a five-digit number is input through the keyboard, write a program to reverse the number.**

```python
n = int(input("Enter a five-digit number: "))

d1 = n % 10
d2 = (n // 10) % 10
d3 = (n // 100) % 10
d4 = (n // 1000) % 10
d5 = n // 10000

reverse = d1*10000 + d2*1000 + d3*100 + d4*10 + d5

print("Reversed Number =", reverse)
```

**In a town, the percentage of men is 52. The percentage of total literacy is 48. If total percentage of literate
men is 35 of the total population, write a program to find the total number of illiterate men and women if
the population of the town is 80,000.**

```python
population = 80000

men = 52 * population / 100
women = population - men

literate_people = 48 * population / 100
literate_men = 35 * population / 100

literate_women = literate_people - literate_men

illiterate_men = men - literate_men
illiterate_women = women - literate_women

print("Illiterate Men =", int(illiterate_men))
print("Illiterate Women =", int(illiterate_women))
```

**A cashier has currency notes of denominations 10, 50 and 100. If the amount to be withdrawn is input
