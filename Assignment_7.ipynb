**Largest of Three Numbers**

```python
def largest_of_three(a,b,c):
    if a > b:
        return a
    elif b > a:
        return b
    elif a > c:
        return a
    elif b > c:
        return b
    else:
        return c
a = int(input())
b = int(input())
c = int(input())
n =largest_of_three(a,b,c)
print(n)
```

**Calculator**

```python
def calculate_percentage(marks):
    total = sum(marks)
    percentage = total / len(marks)
    return percentage

def assign_grade(percentage):
    if percentage >= 90:
        return "A"
    elif percentage >= 75:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 40:
        return "D"
    else:
        return "Fail"

marks = []

for i in range(5):
    mark = int(input("Enter marks of subject {}: ".format(i+1)))
    marks.append(mark)

percentage = calculate_percentage(marks)
grade = assign_grade(percentage)
print("Percentage =", percentage)
print("Grade =", grade)
```

```python
def interest(principal, rate, time):
    si = (principal * rate * time) / 100
    ci = principal * ((1 + rate / 100) ** time) - principal
    return si, round(ci, 2)

principal = float(input("Enter Principal: "))
rate = float(input("Enter Rate: "))
time = int(input("Enter Time: "))

si, ci = interest(principal, rate, time)

print("Simple Interest =", si)
print("Compound Interest =", ci)
```

```python
def is_armstrong(n):
    digits = str(n)
    power = len(digits)
    total = sum(int(digit) ** power for digit in digits)
    return total == n

n = int(input("Enter a number: "))

if is_armstrong(n):
    print(True)
else:
    print(False)
```

```python
def calculator(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        if b == 0:
            return "Cannot divide by zero"
        return a / b
    else:
        return "Invalid Operator"

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
op = input("Enter operator (+, -, *, /): ")

print(calculator(a, b, op))
```

```python
def guess_number_game(target, guesses):
    i = 0
    while i < len(guesses):
        if guesses[i] == target:
            return i + 1
        i += 1
    return "Out of guesses"

print(guess_number_game(42, [10, 20, 30, 42, 50]))
```

```python
def is_leap_year(year):
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False

print(is_leap_year(2024))
```

```python
import random

def rps_game(user_choice):
    choices = ["rock", "paper", "scissors"]
    computer = random.choice(choices)

    if user_choice == computer:
        result = "It's a tie!"
    elif (user_choice == "rock" and computer == "scissors") or \
         (user_choice == "paper" and computer == "rock") or \
         (user_choice == "scissors" and computer == "paper"):
        result = "You win!"
    else:
        result = "Computer wins!"

    return computer, result

computer, result = rps_game("rock")
print("Computer:", computer)
print(result)
```

```python
def bulb_status(switch1, switch2, switch3, correct_switch):
    if correct_switch == 1:
        if switch1:
            return "Bulb is ON"
        else:
            return "Bulb is OFF"

    elif correct_switch == 2:
        if switch2:
            return "Bulb is ON"
        else:
            return "Bulb is OFF"

    elif correct_switch == 3:
        if switch3:
            return "Bulb is ON"
        else:
            return "Bulb is OFF"

print(bulb_status(True, False, False, 1))
```

```python
def ticket_price(age, day, is_3d):
    price = 200

    if age < 12 or age > 60:
        price = price / 2

    if day == "Saturday" or day == "Sunday":
        price += 50

    if is_3d:
        price += 100

    return price

print(ticket_price(8, "Saturday", True))
```

```python
def student_report(maths, science, english):

    if maths < 35 or science < 35 or english < 35:
        return "Fail"

    average = (maths + science + english) / 3

    if average >= 90:
        return "Distinction"
    elif average >= 60:
        return "Second Class"
    elif average >= 35:
        return "Pass Class"

print(student_report(95, 92, 90))
```

```python
def check_booking_conflict(existing_start, existing_end, new_start, new_end):
    if new_start < existing_end and new_end > existing_start:
        return "Conflict! Room already booked"
    else:
        return "No conflict, you can book this slot"

print(check_booking_conflict(10, 12, 11, 13))
```
