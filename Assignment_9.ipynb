```python
#1. Write a function is_anagram (s1, s2) that checks whether two strings are anagrams.
s1 = input()
s2 = input()
def is_anagram(s1,s2):
    if len(s1) != len(s2):
        print('not anagrams')
    else:
        for i in s1:
            if s1.count(i)!=s2.count(i):
                print("not anagrams")
        print('anagrams')       
is_anagram(s1,s2)   
```

```python
#Write a function that takes a list of integers and a target value. Return the indices of two
#numbers whose sum equals the target.
def two_sum(nums,target):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return [i,j]
    return []
nums = [2,7,3,4,9]
target = 9
two_sum(nums,target)
```

```python
#Write a function generate_password(length=8) that generates a random secure password of given
#length.
import random
import string
def generate_password(length=8):
    letters = string.ascii_letters
    password = ""
    for i in range(length):
        password += random.choice(letters)
    return password

pwd = generate_password()
print("Generated Password:", pwd)
```

```python
#Validate Password
def validate_password(password, min_length=8):
    if len(password) < min_length:
        return "Invalid"

    upper = False
    lower = False
    digit = False
    special = False

    for ch in password:
        if ch.isupper():
            upper = True
        elif ch.islower():
            lower = True
        elif ch.isdigit():
            digit = True
        else:
            special = True

    if upper and lower and digit and special:
        return "Valid Password"
    else:
        return "Invalid Password"

print(validate_password("Abc@1234"))
```

```python
#Factorial
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
n=int(input())
print(factorial(n))
```

```python
#Reverse a String
def reverse_string(s):
    if s == "":
        return ""
    return reverse_string(s[1:]) + s[0]
text=input()
print(reverse_string(text))
```

```python
#Fibonacci Number
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))
```

```python
#Decimal to Binary
def decimal_to_binary(n):
    if n == 0:
        return ""
    return decimal_to_binary(n // 2) + str(n % 2)

num = 10
print(decimal_to_binary(num))
```

```python
#Check Palindrome
def palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return palindrome(s[1:-1])

text = input()

if palindrome(text):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Sum of any number of arguments using *args
def find_sum(*args):
    total = 0
    for i in args:
        total += i
    return total

print(find_sum(10, 20, 30, 40))
```

```python
#Return all prime numbers using *args
def prime_numbers(*args):
    prime = []

    for n in args:
        if n > 1:
            for i in range(2, n):
                if n % i == 0:
                    break
            else:
                prime.append(n)

    return prime

print(prime_numbers(2, 3, 4, 5, 6, 7, 8, 11))
```

```python
#Calculate final grades using *args and **kwargs
def student_grade(*marks, **details):
    total = sum(marks)
    average = total / len(marks)

    print("Name:", details["name"])
    print("Average:", average)

    if average >= 90:
        print("Grade: A")
    elif average >= 75:
        print("Grade: B")
    elif average >= 50:
        print("Grade: C")
    else:
        print("Grade: Fail")

student_grade(80, 90, 85, name="Poojitha")
```

```python
#Return the longest word in a sentence
def longest_word(sentence):
    words = sentence.split()
    longest = words[0]

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

print(longest_word("Python programming is interesting"))
```

```python
#Return True if Armstrong number, otherwise False
def armstrong(n):
    temp = n
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** 3
        temp //= 10

    return total == n

print(armstrong(153))
print(armstrong(123))
```

```python
#Return numbers divisible by both 3 and 5
def divisible_numbers(start, end):
    result = []

    for i in range(start, end + 1):
        if i % 3 == 0 and i % 5 == 0:
            result.append(i)

    return result

print(divisible_numbers(1, 100))
```

```python
#Return uppercase, lowercase, and digit count
def count_characters(text):
    upper = 0
    lower = 0
    digit = 0

    for ch in text:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
        elif ch.isdigit():
            digit += 1

    return upper, lower, digit

u, l, d = count_characters("PyThOn123")

print("Uppercase:", u)
print("Lowercase:", l)
print("Digits:", d)
```

```python
#Difference between Global and Local Variables
x = 10      # Global variable

def display():
    y = 20  # Local variable
    print("Inside Function")
    print("Global:", x)
    print("Local:", y)

display()

print("Outside Function")
print("Global:", x)
```

```python
#Modify a Global Variable using global
count = 10

def update():
    global count
    count = count + 5

update()

print("Count =", count)
```
